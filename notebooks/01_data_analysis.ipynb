# NDVI Data Analysis
Load the generated or real NDVI raster and inspect its quality and distribution.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import load_ndvi_geotiff
from src.ndvi import ndvi_statistics, classify_ndvi, class_percentages
ndvi_path = '../data/processed/ndvi.tif'
if not os.path.exists(ndvi_path):
    raise FileNotFoundError('Run `python ../main.py --stage ndvi` first.')
ndvi, profile, crs, transform = load_ndvi_geotiff(ndvi_path)
print(ndvi.shape, crs)
```

```python
pd.Series(ndvi_statistics(ndvi))
```

```python
valid = ndvi[np.isfinite(ndvi)]
sns.histplot(valid, bins=60, kde=True, color='green')
plt.title('NDVI Distribution'); plt.show()
```
