# NDVI Processing
Visualize continuous NDVI and its four vegetation classes.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing import load_ndvi_geotiff
from src.ndvi import classify_ndvi, CLASS_NAMES
ndvi = load_ndvi_geotiff('../data/processed/ndvi.tif')[0]
classes = classify_ndvi(ndvi)
```

```python
plt.figure(figsize=(8,8)); plt.imshow(ndvi, cmap='RdYlGn', vmin=-0.2, vmax=0.8); plt.colorbar(label='NDVI'); plt.axis('off'); plt.show()
```

```python
plt.figure(figsize=(8,8)); plt.imshow(np.where(classes < 0, np.nan, classes), cmap='Greens', vmin=0, vmax=3); cbar=plt.colorbar(ticks=range(4)); cbar.ax.set_yticklabels(CLASS_NAMES); plt.axis('off'); plt.show()
```
