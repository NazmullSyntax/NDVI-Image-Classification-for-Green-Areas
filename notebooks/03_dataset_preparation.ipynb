# Dataset Preparation
Extract 64x64 NDVI patches and inspect spatial train, validation, and test splits.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing import load_ndvi_geotiff
from src.dataset import prepare_dataset
ndvi = load_ndvi_geotiff('../data/processed/ndvi.tif')[0]
data = prepare_dataset(ndvi, patch_size=64, stride=32, out_dir='../data/processed')
[(key, data[key].shape) for key in ('X_train','X_val','X_test','y_train','y_val','y_test')]
```

```python
for name in ('train','val','test'):
    plt.bar(np.arange(4), np.bincount(data[f'y_{name}'], minlength=4), alpha=0.6, label=name)
plt.xticks(range(4), ['C0','C1','C2','C3']); plt.legend(); plt.show()
```

```python
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
for axis, image, label in zip(axes.flat, data['X_train'][:8], data['y_train'][:8]):
    axis.imshow(image[..., 0], cmap='RdYlGn'); axis.set_title(f'Class {label}'); axis.axis('off')
plt.tight_layout(); plt.show()
```
