# CNN Training and Evaluation
Train the four-class NDVI patch classifier and inspect its metrics.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from src.model import build_cnn
data = np.load('../data/processed/ndvi_patches.npz')
model = build_cnn()
model.summary()
```

```python
history = model.fit(data['X_train'], data['y_train'], validation_data=(data['X_val'], data['y_val']), epochs=20, batch_size=32, callbacks=[tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)])
```

```python
os.makedirs('../models', exist_ok=True)
model.save('../models/ndvi_green_space_cnn.keras')
plt.plot(history.history['accuracy'], label='train'); plt.plot(history.history['val_accuracy'], label='validation'); plt.legend(); plt.show()
```
